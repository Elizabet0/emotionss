```python
from PIL import Image
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import os

p = "../data/train"
classes = list(os.listdir(p))

plt.figure(figsize=(15, 8))
for i, em in enumerate(classes):
    em_p = os.path.join("../data/train", em)
    img_path = os.path.join(em_p, os.listdir(em_p)[0])
    img = Image.open(img_path)
    plt.subplot(2, 4, i+1)
    plt.imshow(img ,  cmap='gray')
    plt.title(f"{em}")
    
plt.show()
```

```python
img_array = np.array(img, dtype=np.float32)
print(f"Размер всех картинок: {img_array.shape}")
```

```python
count_train=0
count_test=0

for i, em in enumerate(classes):
    em_p = os.path.join("../data/train", em)
    count_train += len(os.listdir(em_p))

for i, em in enumerate(classes):
    em_p = os.path.join("../data/test", em)
    count_test += len(os.listdir(em_p))
print(f"Трейн: {count_train} ; Тест : {count_test}")
```

```python
classes = []
counts = []
p ="../data/train"

for em in os.listdir(p):
    classes.append(em)
    phots = os.path.join(p, em)
    counts.append(len(os.listdir(phots)))

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)

ax=sns.barplot(x=classes, y=counts)
percentages = [i/count_train*100 for i in counts]

for i in range(len(classes)):
    ax.text(i, counts[i], f'{percentages[i]:.1f}%', ha='center')

plt.title('Train')


plt.subplot(1, 2, 2)

counts2 = []
p = '../data/test'
for em in os.listdir('../data/test'):
    phots = os.path.join(p, em)
    counts2.append(len(os.listdir(phots)))

ax=sns.barplot(x=classes, y=counts2)
percentages2 = [i/count_test*100 for i in counts2]

for i in range(len(classes)):
    ax.text(i, counts2[i], f'{percentages2[i]:.1f}%', ha='center')
plt.title('Test ')


plt.show()
```
